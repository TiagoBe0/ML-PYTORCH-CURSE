# <font color="#004D7F" size=6>1. ¿Qué es `DataLoader`?</font>

Este enfoque simplifica significativamente el manejo de datos en el entrenamiento de modelos de aprendizaje profundo.

- Creación de lotes:
  - Extracción de una porción del conjunto de datos

  - Uso de sintaxis de corte en tensores

- Mejora de la calidad de entrenamiento:
  - Mezclar el conjunto de datos en cada época

  - Evitar lotes idénticos en el bucle de entrenamiento

- Aumento de datos:
  - Generación manual de más variabilidad

  - Común en tareas de imágenes (inclinación, zoom, etc.)

- Uso de `DataLoader`:
  - Simplifica el proceso de manejo de datos

- Creación de `DataLoader`:
  - Requiere lista de pares de muestras (características y target)

  - `DataLoader` usa `len()` para determinar el tamaño del conjunto

  - Parámetro de tamaño de lote
  
  - Configuración recomendada: Usar `shuffle=True` para mezclar muestras en cada época

- Ventajas de `DataLoader`:
  - Mezcla automática de datos entre épocas

  - Facilita la creación de nuevas combinaciones de muestras

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Más información sobre [`DataLoader`](https://pytorch.org/docs/stable/data.html#torch.utils.data.DataLoader)

```python
# Descarga de datos (necesario en Colab): se guardan en ./Datasets
import os, urllib.request

os.makedirs('Datasets', exist_ok=True)
urls = {
    'sonar.csv':
        'https://raw.githubusercontent.com/jbrownlee/Datasets/master/sonar.csv',
}
for nombre, url in urls.items():
    if not os.path.exists(f'Datasets/{nombre}'):
        urllib.request.urlretrieve(url, f'Datasets/{nombre}')
```

```python
import pandas as pd
import torch
from torch.utils.data import DataLoader
from sklearn.preprocessing import LabelEncoder

data = pd.read_csv("Datasets/sonar.csv", header=None)
X = data.iloc[:, 0:60].values
y = data.iloc[:, 60].values

encoder = LabelEncoder()
encoder.fit(y)
y = encoder.transform(y)

X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).reshape(-1, 1)

# DataLoader acepta cualquier secuencia de pares (x, y)
loader = DataLoader(list(zip(X, y)), shuffle=True, batch_size=16)
for X_batch, y_batch in loader:
    print(X_batch, y_batch)
    break
```

Puedes ver en la salida anterior que `X_batch` y `y_batch` son tensores de PyTorch. El `loader` es una instancia de la clase `DataLoader`, que puede funcionar como un iterable. Cada vez que lees de él, obtienes un lote de características y objetivos del conjunto de datos original.

# <font color="#004D7F" size=6>2. `DataLoader` en un Bucle de Entrenamiento</font>

Este ejemplo demuestra cómo integrar `DataLoader` en un flujo de trabajo completo de entrenamiento y evaluación de un modelo de aprendizaje profundo.

- Uso principal de `DataLoader`:
  - Empaquetado del conjunto de entrenamiento

  - Facilita el recorrido en lotes durante el entrenamiento

- Conjunto de test:
  - Puede usar `DataLoader`, pero no es tan crucial
  - El accuracy se calcula sobre todo el conjunto, no por lotes

- Ventajas de este enfoque:
  - Manejo eficiente de lotes durante el entrenamiento

  - Mezcla automática de datos en cada época

  - Facilita la implementación de bucles de entrenamiento estándar

- Consideraciones:
  - El tamaño del lote puede afectar el rendimiento y la velocidad de entrenamiento

  - Ajustar `batch_size` según las capacidades de memoria y las necesidades del modelo

```python
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader
from sklearn.preprocessing import LabelEncoder

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, shuffle=True)

loader = DataLoader(list(zip(X_train, y_train)), shuffle=True, batch_size=16)

model = nn.Sequential(
    nn.Linear(60, 60),
    nn.ReLU(),
    nn.Linear(60, 30),
    nn.ReLU(),
    nn.Linear(30, 1),
    nn.Sigmoid()
)

n_epochs = 200
loss_fn = nn.BCELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)
model.train()
for epoch in range(n_epochs):
    for X_batch, y_batch in loader:
        y_pred = model(X_batch)
        loss = loss_fn(y_pred, y_batch)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

model.eval()
y_pred = model(X_test)
acc = (y_pred.round() == y_test).float().mean()
acc = float(acc)
print("Accuracy: %.2f%%" % (acc*100))
```

# <font color="#004D7F" size=6>3. Iterador de Datos con la clase `Dataset`</font>

Este enfoque proporciona una manera estructurada y flexible de manejar conjuntos de datos en PyTorch, especialmente útil para conjuntos de datos grandes o complejos.

- Clase `Dataset` en PyTorch:
  - Se combina con `DataLoader`

  - Proporciona una interfaz estándar para manejar datos

- Ventajas de usar `Dataset`:
  - Manejo especial de datos antes de obtener muestras

  - Separación clara entre la lógica de acceso a datos y el entrenamiento del modelo
  
  - Facilita la reutilización y el mantenimiento del código

  - Útil para:

    - Lectura de datos desde base de datos o disco

    - Mantener pocas muestras en memoria

    - Procesamiento en tiempo real (ej. aumentación aleatoria en imágenes)

    - Permite cargar solo las muestras necesarias en cada momento

- Implementación de `Dataset`:

  - Derivar de la clase base `Dataset`

  - Agregar dos funciones miembro:

    1. `__len__()`: Retorna el tamaño total del conjunto de datos

    2. `__getitem__()`: Retorna un par (características, objetivo) para un índice dado

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Más información sobre [`Dataset`](https://pytorch.org/docs/stable/data.html#torch.utils.data.Dataset)

```python
from torch.utils.data import Dataset

class SonarDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        features = self.X[idx]
        target = self.y[idx]
        return features, target
```

Este método proporciona una forma más integrada de manejar la división y carga de datos en proyectos PyTorch, mejorando la coherencia y eficiencia del flujo de trabajo.

- Bucle de entrenamiento:
  - Permanece igual después de implementar `Dataset`

  - Conjunto de test aún utiliza tensores de PyTorch directamente

- Alternativa a scikit-learn para división de datos:

  - Función `random_split()` en `torch.utils.data`

  - Trabaja con la clase `Dataset` de PyTorch

- Consideración importante:
  - Modelos PyTorch requieren tensores como entrada, no `Dataset`

- Uso de `default_collate()`:
  - Convierte muestras de `Dataset` a tensores

  - Necesario para alimentar datos al modelo

- Ventajas de este enfoque:
  - Consistencia en el uso de herramientas PyTorch

  - Facilita la integración con otras funcionalidades de PyTorch

- Consideraciones:
  - `random_split()` puede ser más eficiente para grandes conjuntos de datos
  
  - Asegura que los datos permanezcan en el formato correcto para PyTorch


```python
from torch.utils.data import Dataset, DataLoader, random_split, default_collate

data = pd.read_csv("Datasets/sonar.csv", header=None)
X = data.iloc[:, 0:60].values
y = data.iloc[:, 60].values

encoder = LabelEncoder()
encoder.fit(y)
y = encoder.transform(y).reshape(-1, 1)

dataset = SonarDataset(X, y)
trainset, testset = random_split(dataset, [0.7, 0.3])  # fracciones (PyTorch >= 1.13)
loader = DataLoader(trainset, shuffle=True, batch_size=16)

n_epochs = 200
loss_fn = nn.BCELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)
model.train()
for epoch in range(n_epochs):
    for X_batch, y_batch in loader:
        y_pred = model(X_batch)
        loss = loss_fn(y_pred, y_batch)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

X_test, y_test = default_collate(testset)  # apila todas las muestras de test en tensores
model.eval()
y_pred = model(X_test)
acc = (y_pred.round() == y_test).float().mean()
acc = float(acc)
print("Accuracy: %.2f%%" % (acc*100))
```
