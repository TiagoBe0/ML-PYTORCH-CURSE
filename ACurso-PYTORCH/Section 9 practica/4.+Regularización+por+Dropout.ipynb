# <font color="#004D7F" size=6>1. Dropout en Redes Neuronales</font>

- Dropout: Técnica de regularización para redes neuronales
  - Propuesta entre 2012 y 2014

  - Implementada como una capa en la red neuronal

- Funcionamiento durante el entrenamiento:
  - Selecciona aleatoriamente neuronas y las pone a cero

  - Ignora estas neuronas en la pasada hacia adelante

  - No aplica actualizaciones de peso a estas neuronas en la pasada hacia atrás

  - Escala otras neuronas para compensar las eliminadas

- Efectos del Dropout:
  - Mejora el ajuste del modelo a pesar de ser destructivo

  - Reduce la dependencia en patrones específicos (adaptaciones complejas)

  - Fomenta el aprendizaje de múltiples representaciones independientes

- Beneficios:
  - Mejora la generalización del modelo

  - Reduce el sobreajuste a los datos de entrenamiento

  - Aumenta la robustez del modelo

- Mecanismo de acción:
  - Obliga a otras neuronas a compensar por las eliminadas

  - Crea redundancia en la representación de patrones

  - Hace que la red sea menos sensible a patrones específicos

- Consideraciones de uso:
  - La tasa de dropout es un hiperparámetro ajustable

  - Generalmente se aplica después de capas de activación
  
  - Puede requerir ajustes en la tasa de aprendizaje u otros hiperparámetros

# <font color="#004D7F" size=6>2. Regularización con Dropout</font>

- Implementación de Dropout en PyTorch:
  - Uso de la capa `nn.Dropout()`

  - No requiere selección manual de elementos aleatorios

- Funcionamiento de `nn.Dropout()`:
  - Elimina nodos aleatoriamente con probabilidad _p_ durante entrenamiento

  - Escala el tensor resultante por _1/(1-p)_ para mantener el valor promedio

  - Actúa como función identidad durante inferencia

- Consideraciones importantes:
  - Asegurar que el modelo esté en modo inferencia durante evaluación

- Conjunto de datos Sonar:
  - Problema de clasificación binaria (rocas vs minas)
  
  - Datos de entrada numéricos en la misma escala

  - 60 valores de entrada, 1 valor de salida

- Preparación de datos:
  - Estandarización de valores de entrada

- Estructura del modelo base:
  - Dos capas ocultas (60 y 30 unidades)

  - Uso de descenso de gradiente estocástico

  - Tasa de aprendizaje baja y momento

- Evaluación:
  - Uso de scikit-learn

  - Validación cruzada de 10 pliegues
  
  - Resalta diferencias en resultados

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Más información sobre el dataset [`Dropout`](https://pytorch.org/docs/stable/generated/torch.nn.Dropout.html)

```python
# Descarga de datos (necesario en Colab): se guardan en ./Datasets
import os, urllib.request

os.makedirs('Datasets', exist_ok=True)
urls = {
    'sonar.csv':
        'https://raw.githubusercontent.com/jbrownlee/Datasets/master/sonar.csv',
}
for nombre, url in urls.items():
    if not os.path.exists(f'Datasets/{nombre}'):
        urllib.request.urlretrieve(url, f'Datasets/{nombre}')
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedKFold

data = pd.read_csv("Datasets/sonar.csv", header=None)
X = data.iloc[:, 0:60]
y = data.iloc[:, 60]

encoder = LabelEncoder()
encoder.fit(y)
y = encoder.transform(y)

X = torch.tensor(X.values, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).reshape(-1, 1)

class SonarModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(60, 60)
        self.act1 = nn.ReLU()
        self.layer2 = nn.Linear(60, 30)
        self.act2 = nn.ReLU()
        self.output = nn.Linear(30, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.act1(self.layer1(x))
        x = self.act2(self.layer2(x))
        x = self.sigmoid(self.output(x))
        return x

def model_train(model, X_train, y_train, X_val, y_val, n_epochs=300, batch_size=16):
    loss_fn = nn.BCELoss()
    optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9)
    batch_start = torch.arange(0, len(X_train), batch_size)

    model.train()
    for epoch in range(n_epochs):
        for start in batch_start:
            X_batch = X_train[start:start+batch_size]
            y_batch = y_train[start:start+batch_size]
            y_pred = model(X_batch)
            loss = loss_fn(y_pred, y_batch)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    model.eval()
    y_pred = model(X_val)
    acc = (y_pred.round() == y_val).float().mean()
    acc = float(acc)
    return acc

kfold = StratifiedKFold(n_splits=10, shuffle=True)
accuracies = []
for train, test in kfold.split(X, y):
    model = SonarModel()
    acc = model_train(model, X[train], y[train], X[test], y[test])
    print("Accuracy: %.2f" % acc)
    accuracies.append(acc)

mean = np.mean(accuracies)
std = np.std(accuracies)
print("Resultado de línea base: %.2f%% (+/- %.2f%%)" % (mean*100, std*100))
```

# <font color="#004D7F" size=6>3. Dropout en la capa de entrada</font>

Este enfoque proporciona una regularización adicional desde la capa de entrada, potencialmente mejorando la capacidad del modelo para manejar variabilidad en los datos de entrada.

- Ejemplo de implementación:
  - Nueva capa de Dropout entre capa de entrada y primera capa oculta

  - Tasa de Dropout: 20% (1 de cada 5 valores de entrada excluidos aleatoriamente)

- Implicaciones:
  - Puede aumentar la robustez del modelo a ruido en los datos de entrada

  - Potencialmente mejora la generalización del modelo

- Consideraciones:
  - La tasa de Dropout en la entrada puede requerir ajuste según el conjunto de datos
  
  - Puede afectar el tiempo de entrenamiento y la convergencia del modelo

```python
class SonarModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.dropout = nn.Dropout(0.2)  # 20% de las entradas a cero (solo en train)
        self.layer1 = nn.Linear(60, 60)
        self.act1 = nn.ReLU()
        self.layer2 = nn.Linear(60, 30)
        self.act2 = nn.ReLU()
        self.output = nn.Linear(30, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.dropout(x)
        x = self.act1(self.layer1(x))
        x = self.act2(self.layer2(x))
        x = self.sigmoid(self.output(x))
        return x

kfold = StratifiedKFold(n_splits=10, shuffle=True)
accuracies = []
for train, test in kfold.split(X, y):
    model = SonarModel()
    acc = model_train(model, X[train], y[train], X[test], y[test])
    print("Accuracy: %.2f" % acc)
    accuracies.append(acc)

mean = np.mean(accuracies)
std = np.std(accuracies)
print("Dropout en capa de entrada: %.2f%% (+/- %.2f%%)" % (mean*100, std*100))
```

# <font color="#004D7F" size=6>4. Dropout en capas ocultas</font>

Este enfoque es ampliamente utilizado en redes neuronales profundas para mejorar su rendimiento y reducir el sobreajuste.

- Aplicación de Dropout a neuronas ocultas:
  - Más común que en la capa de entrada

  - Se aplica en el cuerpo del modelo de red

- Características clave:
  - Mantiene la estructura general del modelo

  - Aplica Dropout después de cada capa oculta

- Implicaciones:
  - Reduce la dependencia entre neuronas

  - Mejora la capacidad de generalización del modelo

- Consideraciones:
  - La tasa de Dropout puede ajustarse según las necesidades del modelo

  - Puede requerir un aumento en el número de épocas de entrenamiento

- Ventajas:
  - Previene el sobreajuste en capas intermedias
  
  - Fomenta la creación de características más robustas

```python
class SonarModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(60, 60)
        self.act1 = nn.ReLU()
        self.dropout1 = nn.Dropout(0.2)
        self.layer2 = nn.Linear(60, 30)
        self.act2 = nn.ReLU()
        self.dropout2 = nn.Dropout(0.2)
        self.output = nn.Linear(30, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.act1(self.layer1(x))
        x = self.dropout1(x)
        x = self.act2(self.layer2(x))
        x = self.dropout2(x)
        x = self.sigmoid(self.output(x))
        return x
    
kfold = StratifiedKFold(n_splits=10, shuffle=True)
accuracies = []
for train, test in kfold.split(X, y):
    model = SonarModel()
    acc = model_train(model, X[train], y[train], X[test], y[test])
    print("Accuracy: %.2f" % acc)
    accuracies.append(acc)

mean = np.mean(accuracies)
std = np.std(accuracies)
print("Dropout en capas ocultas: %.2f%% (+/- %.2f%%)" % (mean*100, std*100))
```

# <font color="#004D7F" size=6>5. Dropout en modo de evaluación</font>

- Comportamiento de Dropout después del entrenamiento:
  - No tiene efecto durante la inferencia

  - Funciona como una función de identidad en modo evaluación

- Importancia de `model.eval()`:
  - Debe usarse antes de evaluar el modelo

  - Asegura que Dropout no afecte las predicciones

- Funcionamiento en PyTorch:
  - La capa Dropout detecta automáticamente el modo de evaluación

  - No requiere modificaciones manuales del código para inferencia

- Razones para desactivar Dropout en inferencia:
  - Proporcionar toda la información posible al modelo

  - Utilizar la capacidad completa de la red entrenada

- Objetivo de Dropout durante el entrenamiento:
  - Asegurar que la red aprenda suficientes pistas sobre la entrada
  
  - Evitar dependencia de fenómenos raros en los datos

- Consideraciones importantes:
  - Asegurarse de alternar entre `model.train()` y `model.eval()` según sea necesario

  - Mantener consistencia en el uso de estos modos durante entrenamiento y evaluación

- Impacto en el rendimiento:
  - Mejora la generalización del modelo

  - Puede resultar en predicciones más robustas en datos no vistos

- Recordatorio:
  - Siempre verificar que el modelo esté en el modo correcto antes de entrenar o evaluar

  - La omisión de `model.eval()` puede llevar a resultados inconsistentes en la inferencia

# <font color="#004D7F" size=6>6. Consejos para usar Dropout</font>

El artículo original sobre Dropout proporciona resultados experimentales en una serie de problemas estándar de aprendizaje automático. Como resultado, ofrecen una serie de heurísticas útiles a considerar al usar Dropout en la práctica:

- Generalmente, usa un valor pequeño de dropout del 20% al 50% de las neuronas, siendo el 20% un buen punto de partida. 

    - Una probabilidad demasiado baja tiene un efecto mínimo, y un valor demasiado alto resulta en un aprendizaje insuficiente por parte de la red.

- Usa una red más grande. 

    - Es probable que obtengas un mejor rendimiento cuando se usa Dropout en una red más grande, lo que le da al modelo más oportunidades para aprender representaciones independientes.

- Aplica Dropout tanto en las unidades de entrada (visibles) como en las unidades ocultas. 

    - La aplicación de Dropout en cada capa de la red ha mostrado buenos resultados.

- Usa una tasa de aprendizaje grande con decaimiento o un gran valor de momentum. 

    - Aumenta tu tasa de aprendizaje por un factor de 10 a 100, o usa un valor de momentum alto de 0.9 o 0.99. 

    - Esto puede parecer contrario a la intuición, pero los autores argumentan que Dropout introduce mucho ruido y esto puede ayudar a acelerar el aprendizaje.

- Restringe el tamaño de los pesos de la red. Una tasa de aprendizaje grande puede resultar en pesos de red muy grandes. 

    - Imponer una restricción sobre el tamaño de los pesos de la red, como la regularización **max-norm** con un tamaño de 4 o 5, ha demostrado mejorar los resultados.
